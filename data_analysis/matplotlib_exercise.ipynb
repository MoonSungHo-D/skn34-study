- **객체 지향 인터페이스**
    - `pyplot` 인터페이스는 간단한 그래프에 적합하지만, 보다 복잡한 그래프를 그리거나 여러 개의 그래프를 한 화면에 그릴 때는 객체 지향 방식이 더 유용하다.
    - 이 방식은 복잡한 그래프를 그리거나, 여러 개의 서브플롯을 만들 때 유리하다.
    - Matplotlib의 객체 지향 방식에서는 Figure와 Axes를 직접 생성하고 관리할 수 있다. 아래는 객체 지향 방식으로 그래프를 그리는 예제이다.

```python
import matplotlib.pyplot as plt

# Figure와 Axes 객체 생성
fig, ax = plt.subplots()

# Axes 객체에서 그래프 그리기
ax.plot([1, 2, 3, 4], [10, 20, 25, 30])
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_title('OOP-Graph')

plt.show()
```

pyplot 방식 : 하나만 간단하게 그림  
figure, axes 객체 방식 : subplots 그릴때 유용

## 1-2. 선 그래프

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temperatures = [22, 24, 19, 23, 25, 20, 18]

plt.plot(
    days,
    temperatures, 
    marker= 'o',
    linestyle= '-',
    color = 'b'
    )

plt.title('Temperatures Variation Over a Week')
plt.xlabel("Day")
plt.ylabel("Temperatures")
plt.grid(True) # 격자 생성
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y1 = [10, 15, 20, 25, 30]
y2 = [5, 10, 15, 20, 25]

plt.plot(x, y1, label="Dataset 1", marker= 'o')
plt.plot(x, y2, label="Dataset 2", marker= 's')

plt.title('Customized Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.grid(True)
plt.show()
```

## 1-3. 막대 그래프

```python
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [150, 200, 180, 220, 170]

# 막대 그래프 그리기
plt.plot(months, sales,"r--",marker= 'o')
plt.bar(months, sales, color='skyblue')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales Amount')

for i, month in enumerate(months):
    plt.text(month, sales[i], sales[i])

plt.show()
```

```python
# plt.barh() : 가로 막대 그래프
# 데이터 준비
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [150, 200, 180, 220, 170]

# 막대 그래프 그리기
plt.barh(months, sales, color='skyblue')
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales Amount')
plt.show()
```

### 1-5-2. 히스토그램 그리기

```python
import matplotlib.pyplot as plt
import numpy as np

# 데이터 준비
heights = np.random.normal(170, 5, 200)  # 평균 170cm, 표준편차 5cm인 데이터 200개 생성

# 히스토그램 그리기
plt.hist(heights, bins=10, color='green', edgecolor='black')
plt.title('Height Distribution of Students')
plt.xlabel('Height (cm)')
plt.ylabel('Number of Students')
plt.show()
```

## 1-6. 산점도



```python

# 데이터 준비
np.random.seed(0)
study_hours = np.random.normal(5, 1.5, 100)  # 평균 5시간, 표준편차 1.5시간
scores = study_hours * 10 + np.random.normal(0, 5, 100)  # 공부 시간에 따른 점수

# 산점도 그리기
plt.scatter(study_hours, scores, color='purple', alpha=0.6)
plt.title('Relationship between Study Hours and Exam Scores')
plt.xlabel('Study Hours')
plt.ylabel('Exam Score')
plt.show()
```
