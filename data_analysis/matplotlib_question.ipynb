## [실습 1] 다양한 그래프 활용 1

- 주어진 데이터에 대해 선 그래프, 막대 그래프, 히스토그램, 산점도를 한꺼번에 그려보세요.


```python
import matplotlib.pyplot as plt
import numpy as np

np.random.seed(0)

# 데이터 준비
x = np.arange(1, 11)
y = np.random.randint(10, 100, 10)

data = np.random.normal(50, 15, 100)
```

```python
plt.figure(figsize=(10,10))

# plot 
plt.subplot(2,2,1)
plt.plot(x,y)
plt.title('plot')

# barplot 
plt.subplot(2,2,2)
plt.bar(x,y)
plt.title('barplot')

# histogram
plt.subplot(2,2,3)
plt.hist(data)
plt.title('histo')

# scatter
plt.subplot(2,2,4)
plt.scatter(x,y)
plt.title('scatter')
```

## [실습 2] 다양한 그래프 활용 2

- 다음 데이터를 가지고 문제를 해결하세요.
  
- **월별 평균 기온 (°C)**
    - `months`: ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']
    - `avg_temperatures`: [-1, 0, 5, 10, 15, 20, 25, 23, 19, 12, 5, 0]
- **월별 강수량 (mm)**
    - `rainfall`: [50, 40, 60, 80, 100, 120, 150, 140, 110, 90, 70, 60]
- **학생들의 시험 점수**
    - `math_scores`: [85, 90, 88, 76, 95, 89, 72, 84, 91, 87]
    - `english_scores`: [78, 82, 85, 80, 88, 90, 84, 79, 85, 83]
    
    1. 월별 평균 기온을 시각화하는 선 그래프를 그려보세요.
        1. 월별 강수량을 시각화하는 막대 그래프를 그려보세요.
        2. 수학과 영어 점수의 분포를 비교하는 박스플롯 그래프를 그려보세요.
        3. 수학 점수 분포를 분석하는 히스토그램을 그려보세요.
        4. 수학 점수와 영어 점수의 상관관계를 분석하는 산점도를 그려보세요.
        5. 1~5의 모든 그래프를 하나의 Figure에 배치해 보세요.

```python
import pandas as pd
import numpy as np


months = ['January', 'February', 'March', 'April', 'May', 'June', 
          'July', 'August', 'September', 'October', 'November', 'December']
avg_temperatures = [-1, 0, 5, 10, 15, 20, 25, 23, 19, 12, 5, 0]
rainfall = [50, 40, 60, 80, 100, 120, 150, 140, 110, 90, 70, 60]
math_scores = [85, 90, 88, 76, 95, 89, 72, 84, 91, 87]
english_scores = [78, 82, 85, 80, 88, 90, 84, 79, 85, 83]

data_weather = {
    'months' : ['January', 'February', 'March', 'April', 'May', 'June', 
          'July', 'August', 'September', 'October', 'November', 'December'],
    'avg_temperatures' : [-1, 0, 5, 10, 15, 20, 25, 23, 19, 12, 5, 0],
    "rainfall": [50, 40, 60, 80, 100, 120, 150, 140, 110, 90, 70, 60]
}
df_weather = pd.DataFrame(data_weather)
df_weather

data_score = {
    'math_scores' : [85, 90, 88, 76, 95, 89, 72, 84, 91, 87],
    'english_scores' : [78, 82, 85, 80, 88, 90, 84, 79, 85, 83]
}

df_score = pd.DataFrame(data_score)
```

```python
# 1. 월별 평균 기온을 시각화하는 선 그래프를 그려보세요.
plt.plot(df_weather["months"], df_weather['avg_temperatures'])
plt.title('AVERAGE_MONTHLY_TEMP')
plt.xlabel('Months')
plt.ylabel('AVG-temperatures')
plt.xticks(rotation=45)
```

```python
# 2. 월별 강수량을 시각화하는 막대 그래프를 그려보세요.

plt.bar(df_weather["months"], df_weather['rainfall'])
plt.title('MONTHLY_Rainfall')
plt.xlabel('Months')
plt.ylabel('Rainfall')
plt.xticks(rotation=45)
```

```python
# 3. 수학과 영어 점수의 분포를 비교하는 박스플롯 그래프를 그려보세요.
plt.boxplot([df_score["math_scores"],df_score["english_scores"]])
plt.title('Math VS English')
```
