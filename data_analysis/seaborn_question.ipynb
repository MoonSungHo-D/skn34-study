```python
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.font_manager as format
# 한글 폰트 설정
plt.rcParams['font.family'] = 'Malgun Gothic'

# 마이너스 기호 꺠짐 방지
plt.rcParams["axes.unicode_minus"] = False

# 데이터셋 로드 
titanic = sns.load_dataset('titanic')
titanic
```



- 문제 1: 생존자와 사망자 수 시각화
    - Titanic 데이터셋의 `survived` 열을 사용하여 생존자와 사망자의 수를 막대 그래프로 시각화하세요.
    - 그래프에 제목, 축 레이블을 추가하고 색상을 설정하세요.

```python

# plt.bar(titanic['survived'],titanic['survived'].count())

plt.figure(figsize=(10,6))

sns.countplot(x='survived', data= titanic, palette='muted')
plt.title('생존자와 사망자의 수 그래프')
plt.xlabel('생존여부(0= 사망, 1= 생존)')
plt.ylabel('생존자 수')
```

- 문제 2: 성별에 따른 생존율 비교
    - Titanic 데이터셋의 `sex`와 `survived` 열을 사용하여 성별에 따른 생존율을 막대 그래프로 시각화하세요.
    - 성별 생존율을 비교하고 제목, 축 레이블을 추가하세요.

```python
plt.figure(figsize=(10,6))

sns.barplot(x='sex', y= 'survived', data= titanic, palette='coolwarm')
plt.title('성별에 따른 생존율')
plt.xlabel('성별')
plt.ylabel('생존율')
plt.legend()
plt.show()
```

- 문제 3: 클래스와 성별에 따른 생존율 비교
    - `class`를 x축으로 설정하고, 성별에 따라 색상을 구분하여 그래프를 생성하세요.
    - Titanic 데이터셋의 `class`, `sex`, `survived` 열을 사용하여 그룹별 생존율을 시각화하세요.
    - 그래프에 제목, 축 레이블, 범례를 추가하세요.

```python
plt.figure(figsize=(10,6))


sns.barplot(x='class', y= 'survived', data= titanic, palette='coolwarm',hue='sex')
plt.title('클래스에 따른 생존율')
plt.xlabel('클래스')
plt.ylabel('생존율')
plt.show()
```

- 문제 4: 나이 분포 분석
    - Titanic 데이터셋의 `age` 열을 사용하여 승객들의 나이 분포를 히스토그램과 KDE 그래프로 시각화하세요.
    - 데이터에서 결측치를 제거하고 그래프를 그리세요.

```python
plt.figure(figsize=(10,6))

sns.histplot(titanic['age'].dropna(),kde = True, bins= 20,color='pink')
plt.title('나이분포 분석')
plt.xlabel('나이')
plt.ylabel('수')
plt.show()
```

- 문제 5: 나이와 요금의 관계 분석
    - Titanic 데이터셋의 `age`, `fare`, `survived` 열을 사용하여 나이와 요금 간의 관계를 산점도로 시각화하세요.
    - 생존 여부(`survived`)에 따라 색상을 다르게 설정하세요.
    - 그래프에 제목, 축 레이블을 추가하세요.

```python
plt.figure(figsize=(10,6))


sns.scatterplot(x= 'age', y='fare', data=titanic,hue='survived')
plt.title('나이와 요금의 관계 산점도')
plt.xlabel('나이')
plt.ylabel('요금')
plt.show()
```
