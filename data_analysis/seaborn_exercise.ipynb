### 2-2. 선 그래프 (lineplot)

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np  # numpy 직접 가져오기

# 예제 데이터 생성
dates = pd.date_range('2021-01-01', periods=100)
prices = pd.Series(range(100)) + np.random.randn(100).cumsum()  # np.random으로 변경
stock_data = pd.DataFrame({'Date': dates, 'Price': prices})

sns.set_theme(style= "white")

# 선 그래프 그리기
sns.lineplot(x='Date', y='Price', data=stock_data)
plt.title('Stock Price Over Time')
plt.xlabel('Date')
plt.ylabel('Price')
plt.xticks(rotation=45)
plt.show()
```

### 2-3. 막대 그래프(barplot)

```python
# 데이터셋 로드
tips = sns.load_dataset('tips')
tips.head()
```

```python
# 막대 그래프 그리기
sns.barplot(x='day', y='tip', data=tips, estimator=sum)
plt.title('Total Tips by Day')
plt.xlabel('Day of Week')
plt.ylabel('Total Tip Amount')
plt.show()
```

## 2-4. 히스토그램과 KDE (Kernel Density Estimation)

### 2-4-1. KDE란

<aside>
💡

히스토그램은 데이터의 분포를 구간별 빈도로 나타내는 그래프이고, KDE는 히스토그램과 마찬가지로 데이터의 분포를 나타내지만 부드러운 곡선으로 추정하여 나타낸다.

</aside>

```python
# 데이터셋 로드
tips = sns.load_dataset('tips')

# 히스토그램과 KDE 그리기
sns.histplot(tips['total_bill'], kde=True, bins=20)
plt.title('Distribution of Total Bill')
plt.xlabel('Total Bill')
plt.ylabel('Frequency')
plt.show()
```

## 2-5. 박스플롯과 바이올린 플롯

### 2-5-1. 바이올린 플롯이란

<aside>
💡

박스플롯은 데이터의 사분위수, 중앙값, 이상치를 시각화하는 그래프이고, 바이올린 플롯은 박스플롯과 KDE를 결합하여 데이터의 분포와 사분위수를 동시에 보여준다.

</aside>

```python

# 데이터셋 로드
tips = sns.load_dataset('tips')

# 박스플롯 그리기
sns.boxplot(x='sex', y='tip', data=tips)
plt.title('Tip Amount by Sex')
plt.xlabel('Sex')
plt.ylabel('Tip Amount')
plt.show()
```

```python
# 바이올린 플롯 그리기
sns.violinplot(x='sex', y='tip', data=tips)
plt.title('Tip Amount by Sex')
plt.xlabel('Sex')
plt.ylabel('Tip Amount')
plt.show()
```

### 2-7. 산점도 (scatterplot)

```python
import seaborn as sns
import matplotlib.pyplot as plt

# 데이터셋 로드
tips = sns.load_dataset('tips')

# 산점도 그리기
sns.scatterplot(x='total_bill', y='tip', data=tips, hue='time')
plt.title('Total Bill vs Tip')
plt.xlabel('Total Bill')
plt.ylabel('Tip')
plt.show()
```
