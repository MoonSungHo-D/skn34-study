- **결측값 제거 (`dropna()`)**
    - 행 또는 열을 기준으로 제거할 수 있다.

````python
import numpy as np
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David', 'Eve'],
    'Age': [25, 30, np.nan, 40, 28],
    'City': ['New York', 'Los Angeles', 'Chicago', 'Houston', 'San Francisco'],
    'Salary' : [70000, 80000, 120000, 90000,  np.nan]
}

df = pd.DataFrame(data)
df
````

````python
# 결측값이 있는 행 제거
df_cleaned = df.dropna()
print(df_cleaned)
````

````python
# 결측값을 0으로 대체
df_filled = df.fillna(0)
print(df_filled)

# 결측값을 평균값으로 대체
df['Age'] = df['Age'].fillna(df['Age'].mean())
print(df)

````

````python
df['Salary'] = df['Salary'].fillna(df['Salary'].median())
print(df)
````

### **4-6-2. 데이터 조작**

- **`merge()` 함수**
    - pandas의 `merge()` 함수는 SQL의 JOIN과 유사하게 두 개의 DataFrame을 특정 열을 기준으로 결합한다.
    - 다양한 방법으로 조인할 수 있으며, `left`, `right`, `inner`, `outer` 조인이 가능하다.

````python
# 두 개의 데이터프레임 생성
df1 = pd.DataFrame({'CustomerID': [1, 2, 3], 'Product': ['Laptop', 'Mouse', 'Keyboard']})
df2 = pd.DataFrame({'CustomerID': [2, 3, 4], 'Location': ['USA', 'Canada', 'Mexico']})

# CustomerID를 기준으로 두 DataFrame 결합 (inner join)
df_merged = pd.merge(df1, df2, on= "CustomerID", how="inner")
print(df_merged)
print("--------")
# left join
df_merged = pd.merge(df1, df2, on= "CustomerID", how="left")
print(df_merged)
print("--------")


# outer join
df_merged = pd.merge(df1, df2, on= "CustomerID", how="outer")
print(df_merged)

````

- **`concat()` 함수**
    - `concat()` 함수는 여러 DataFrame을 위아래로 또는 좌우로 결합할 때 사용된다.
        
        이때 행을 추가하는 경우 `axis=0`, 열을 추가하는 경우 `axis=1`을 사용한다.

````python
# 두 개의 데이터프레임을 세로로 연결
df1 = pd.DataFrame({'A': ['A0', 'A1'], 'B': ['B0', 'B1']})
df2 = pd.DataFrame({'A': ['A2', 'A3'], 'B': ['B2', 'B3']})
print(df1)
print(df2)
print('--------')

# 행으로 추가
df_concat = pd.concat([df1, df2], axis= 0)
print(df_concat)
print('--------')

# 인덱스 맞추기
df_concat = pd.concat([df1, df2], axis= 0,ignore_index=True)
print(df_concat)
print('--------')

# 열로 추가
df_concat = pd.concat([df1, df2], axis= 1)
print(df_concat)
````

- **`join()` 함수**
    - `join()`은 두 개의 DataFrame을 인덱스를 기준으로 결합할 때 사용된다.
    - 주로 인덱스가 동일한 경우에 사용된다.

### **4-6-3. 데이터 피벗 및 변형**

- **`pivot()` 함수**
    - `pivot()` 함수는 데이터프레임을 재구조화 하여 행, 열, 값으로 표현할 수 있는 형식으로 변환한다.
    - 주로 `index`, `columns`, `values`를 지정하여 사용한다.

````python
data = {'Date': ['2024-01-01', '2024-01-01', '2024-01-02'],
        'City': ['New York', 'Los Angeles', 'New York'],
        'Temperature': [55, 68, 52]}

df_weather = pd.DataFrame(data)
print(df_weather)
print('----')
# pivot을 이용해 날짜별 도시 온도 정리
df_pivot = df_weather.pivot(index='Date', columns='City', values='Temperature')
df_pivot
````

- **`pivot_table()` 함수**
    - `pivot_table()` 함수는 그룹화된 데이터를 요약하여 피벗 테이블로 변환할 수 있다.
        
        `aggfunc` 옵션을 사용하여 합계, 평균 등의 연산을 할 수 있다.

````python
sales_data = {
    'Product': ['A', 'B', 'A', 'B'],
    'Sales': [100, 150, 200, 250],
    'Region': ['North', 'South', 'North', 'South']
}


df_sales = pd.DataFrame(sales_data)
print(df_sales)
print('---')


# pivot_table을 이용해 지역별, 상품별 총 판매량 요약
df_pivot_table = df_sales.pivot_table(values='Sales', index='Region', 
                                      columns='Product', aggfunc='sum')
print(df_pivot_table)
````

- **`melt()` 함수**
    - `melt()` 함수는 데이터프레임을 넓은 형식에서 긴 형식으로 변환하는 데 사용된다.
        
        이는 데이터를 열 기준으로 "녹여" 변수와 값 형태로 변환다.
        

````python
df_wide = pd.DataFrame({
    'ID': [1, 2, 3],
    'Math': [85, 90, 88],
    'English': [78, 85, 90]
})
print(df_wide)
print('-----')
# melt를 이용해 긴 형식으로 변환
df_melt = pd.melt(df_wide, id_vars=['ID'], value_vars=['Math', 'English'],
                  var_name='Subject', value_name='Score')
print(df_melt)
````

### **4-7-1. 데이터 그룹화**

- **`groupby()` 함수**
    - 데이터를 특정 기준에 따라 그룹화하여 집계 연산을 수행한다.
        
        이를 통해 데이터의 패턴을 발견하거나 요약 통계를 얻는 데 매우 유용하다.
        
    - 기본적인 그룹화`groupby()`는 주어진 열을 기준으로 데이터를 그룹화하고, 그 결과에 대해 다양한 집계 연산을 적용할 수 있다.

````python

data = {
    'Department': ['HR', 'IT', 'HR', 'Finance', 'IT', 'Finance'],
    'Salary': [50000, 60000, 52000, 80000, 65000, 90000]
}
df = pd.DataFrame(data)

# 부서별로 평균 급여 계산
grouped = df.groupby('Department')['Salary'].mean()
print(grouped)
````

````python
data = {
    'Department': ['HR', 'IT', 'HR', 'Finance', 'IT', 'Finance'],
    'Location': ['New York', 'New York', 'Chicago', 'Chicago', 'New York', 'Chicago'],
    'Salary': [50000, 60000, 52000, 80000, 65000, 90000]
}
df = pd.DataFrame(data)

# 부서와 위치별로 평균급여

grouped = df.groupby(["Department","Location"])["Salary"].mean()
print(grouped)
````

### **4-7-2. 데이터 집계 함수**


- **집계 연산 적용**
    - **`agg()` 함수**는 그룹화된 데이터에 대해 원하는 집계 연산을 적용하여 요약 통계를 얻을 수 있다.

````python
data = {
    'Department': ['HR', 'IT', 'HR', 'Finance', 'IT', 'Finance'],
    'Location': ['New York', 'New York', 'Chicago', 'Chicago', 'New York', 'Chicago'],
    'Salary': [50000, 60000, 52000, 80000, 65000, 90000]
}
df = pd.DataFrame(data)
df
````

````python
# department 를 기준으로 salary의 총합, 평균, 갯수
aggregated = df.groupby('Department').agg({
    "Salary" : ["sum","mean","count"]
})

aggregated
````

````python
df.groupby("Department")["Salary"].agg(["sum","mean","min","max"])
````

### **4-7-3. 그룹화된 데이터 필터링**

- **`filter()` 함수**
    - 그룹별로 조건을 검사하여 해당 조건을 만족하는 그룹만 반환한다.
        
        그룹화된 데이터에서 조건을 걸어 필요한 그룹만 선택할 수 있다.

````python
# 평균 급여가 60000 이상인 부서만 선택
filtered = df.groupby('Department').filter(lambda x: x['Salary'].mean() >= 60000)
print(filtered)
````

### **4-7-4. 그룹화된 데이터 재구성**


- **`unstack()` 함수**
    - 데이터를 한 차원 더 넓은 형식으로 변환한다.

````python
unstacked = df.groupby(['Department', 'Location'])['Salary'].mean()
print("------unstack 없을때------")
print(unstacked)

unstacked = df.groupby(['Department', 'Location'])['Salary'].mean().unstack()
print("------unstack 있을떄------")
print(unstacked)
````

- **`stack()` 함수**
    - 데이터를 한 차원 더 긴 형식으로 변환한다.

````python
stacked = unstacked.stack()
stacked
````

- **`reset_index()` 함수**
    - 인덱스를 리셋하여 기존 인덱스를 새로운 열로 변환한다.

````python
df_reset = df.groupby('Department')['Salary'].mean().reset_index()
print(df_reset)
````

### **4-8-2. 시간 시계열 데이터 처리**



- **날짜 형식으로 변환**
    - 시계열 데이터를 처리하기 위해서는 날짜 정보를 **datetime** 형식으로 변환해야 한다. `pd.to_datetime()` 함수를 사용하여 문자열 형식의 날짜를 datetime 객체로 변환할 수 있다.

````python
# 날짜 데이터를 datetime 형식으로 변환
df['Date'] = pd.to_datetime(df['Month'] + ' 2026')
print(df)
````

````python
# 월별 데이터를 날짜로 변환 후 집계 예시
df.set_index('Date', inplace=True)
df
````

````python
monthly_sales = df.resample('ME').sum()  # 'ME'은 월별 집계를 의미
print(monthly_sales)
````

````python
# 시계열 데이터 시각화
monthly_sales.plot(title='Monthly Sales Over Time')
plt.show()
````

- **이동 평균 (Rolling Mean)**
    - 이동 평균은 시계열 데이터를 부드럽게 만들어주는 중요한 기법 중 하나다. `rolling()` 메서드를 사용하여 이동 평균을 구할 수 있다.

````python
# 3개월 이동 평균 계산 및 시각화
df['RollingMean'] = df['Sales'].rolling(window=3).mean()
print(df)
df[['Sales', 'RollingMean']].plot(title='Sales and 3-Month Rolling Average')
plt.show()
````
