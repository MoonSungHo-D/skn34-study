## Encoder-Only 모델 : 네이버 영화리뷰로 학습한 감성분석 모델

```python
from transformers import pipeline   # hugging-face 모델을 간단히 사용할 수 있는 함수
```

```python
classifier = pipeline('text-classification', model = 'daekeun-ml/koelectra-small-v3-nsmc')
```

```python
def analyze_sentiment(texts):
    results = classifier(texts)
    
    label_map = {"0": '부정', "1": '긍정'}
    
    for text, result in zip(texts, results):
        print(f'입력 문장 : {text}')
        print(f'모델 결과 : {result}')
        print(f'감성 : {label_map[result['label']]}')
        print(f'확률 : {round(result["score"], 4)}')
        print()
```

```python
texts = ['진짜 이런 영화는 왜 만드는지 모르겠다', '가족들이랑 편하게 즐기기 좋은 킬링타임 오락액션영화 ㅋㅋ 전편보다 스케일이 커져서 시원시원했고 빌런 캐릭터랑 액션도 꽤 좋았음 ㅋㅋㅋ', ' 1편은 그냥 b급 코미디영화다 라는 생각으로 봤는데 2편이 나올줄은 몰랐다 근데 3편은 절대 안나올듯 수중에서 폭탄이터졌는데 왜 살아남은거임? 감독 문과임?','영화는 잘못이 없다. 이럴 걸 예상했음에도 포인트 적립하겠다고 시간 내서 본 나에게 책임이 있다.']

analyze_sentiment(texts)
```

## Decoder-only 모델 : 언어생성

```python
generater = pipeline('text-generation', model = 'Qwen/Qwen2.5-0.5B-Instruct')
```

```python
# 입력 문장을 받아 자연스럽게 이어서 생성하는 함수
def generation_text(text):
    messages = [    # 모델에게 전달할 입력 메시지
        {
            'role': 'user',     # 사용자 메시지
            'content': f"다음 문장을 자연스럽게 이어서 작성해줘. {text}"
        }
    ]
    
    result = generater(
        messages, 
        max_new_tokens = 100, 
        do_sample = True,
        temperature = 0.7
    )
    
    return result[0]['generated_text'][-1]['content'], result
```

```python
text = '오늘 학원에서 수업을 듣다가'
print(generation_text(text))
```

```python
texts = [
    '인공지능이 발전하게 되면 인간을',
    '인간이 살아가는 이유는',
    '한국 사람이 똑똑한 이유는',
    '친일파는',
    '내가 사는 이유는'
]

for text in texts:
    print(f'입력 : {text}')
    print(f'생성 : {generation_text(text)}' )
    print()
```
