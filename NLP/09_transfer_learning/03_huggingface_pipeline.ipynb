```python
# 한국어 -> 영어 번역

translator = pipeline('translation', model= 'Helsinki-NLP/opus-mt-ko-en')
translator
```

```python
# zero-shot Classification
zero_shot_clf = pipeline('zero-shot-classification')
```

```python
zero_shot_clf(
    'This is a course about the tranformers library.',
    candidate_labels = ['education', 'politics', 'business']
)
```

```python
ko_zero_shot_clf = pipeline('zero-chot-classification', model = 'joeddav/xlm-roberta-large-xnli')
```

```python
ko_text = '2026년 당신은 얼마나 잘 살고있나요?'

candidates= ['사회', '경제','건강','정치']

ko_zero_shot_clf(ko_text, candidates)
```

```python
# 텍스트 생성
text_generator = pipeline('text-generation', model= 'distilgpt2')
```

```python
text_generator(
    'In the course, we will teach you how to',
    max_len= 30,                # 생성 결과 최대 토큰 길이
    num_return_sequences = 2    # 서로 다른 생성 결과 2개
)
```
