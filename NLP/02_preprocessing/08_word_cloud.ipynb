```python
import numpy as np
import pandas as pd
from konlpy.tag import Okt          # 한국어 형태소 분석기
from collections import Counter     # 빈도수 계산
from wordcloud import WordCloud     # 워드클라우드 생성 
import matplotlib.pyplot as plt 
```

```python
spam_df = pd.read_csv('data/sms_spam.csv')
spam_df
```

```python
corpus = spam_df['CN']
corpus
```

```python
from tqdm.auto import tqdm

okt = Okt()
nouns = []

for corpus_temp in tqdm(corpus):            # 전체 코퍼스를 한 문장씩 순회
    nouns.extend(okt.nouns(corpus_temp))    # 명사 리스트를 추출
    
nouns
```

```python
import re

for line in lines:
    if '] [' in line:
        msg = line.split('] ', 2)[2]

        msg = re.sub(r'[ㅋㅎㅠㅜ]+', '', msg)          # ㅋㅋ,ㅎㅎ,ㅠㅠ,ㅜㅜ 제거
        msg = re.sub(r'사진\n|이모티콘\n', '', msg)     # 사진, 이모티콘 제거
        msg = msg.replace('삭제된 메시지입니다', '')
        msg = msg.replace('\n', ' ')

        text += msg
```
