````python
import numpy as np
import pandas as pd
import os, glob

path = './data/OpinosisDataset1.0/topics'
all_files = glob.glob(os.path.join(path, '*.data'))

filename_list = []  # 파일명
opinions_list = []  # 파일 내용(의견)

for file_ in all_files:
    df = pd.read_table(
        file_,              # 경로
        header= None,       # 첫 줄은 컬럼명으로 사용안함
        index_col= None,    # 인덱스 컬럼 지정 안함(기본)
        encoding= 'latin1'  # 특수문자 깨짐방지
    )
    # display(df)
    
    filename = os.path.basename(file_)  # 파일명 추출
    filename = filename.split('.')[0]   # 확장자 제거
    filename_list.append(filename)      # 파일명 리스트 추가
    
    opinions = df.to_string(index=False, header=False) # 인덱스/ 헤더 제거
    opinions_list.append(opinions)
````

````python
document_df = pd.DataFrame({
    'filename' : filename_list,
    'opinions' : opinions_list
})

document_df
````

### 특성 벡터화 및 전처리
- TfidfVectorizer 이용한 벡터화
- 불용어 처리
- ngram 설정
- 어근 분리(Lemmoatization)

````python
import nltk

nltk.download('wordnet')        # 대형 의미 사전
````

````python
from nltk.stem import WordNetLemmatizer     # 표제어(lemma) 추출 도구

lemmatizer = WordNetLemmatizer()
print(lemmatizer.lemmatize('running', pos= 'v'))    # 동사로 간주해 표제어(사전 대표 기본형) 변환
print(lemmatizer.lemmatize('ran', pos= 'v'))        
````

````python
# 파이썬에서 사용하는 문장부호 목록
import string
string.punctuation
````

````python
# 텍스트를 정리하고, 표제어(동사 기준)로 변환하는 함수
def lemmatizer(text):
    text = text.lower()     # 소문자 변환
    
    punc_rem_dict = dict((ord(ch), None) for ch in string.punctuation)  # 문장부포 제거용 치환 딕셔너리
    text = text.translate(punc_rem_dict)    # text 에서 문장부호에 해당하는 것들은 None 
    
    tokens = nltk.word_tokenize(text)
    
    lemmatizer = WordNetLemmatizer()
    return [lemmatizer.lemmatize(token, pos= 'v') for token in tokens]

lemmatizer('The Matrix is everywhere its all around us, here even in this room. \
You can see it out your window or on your television. \
You feel it when you go to work, or go to church or pay your taxes.')
````

````python
# Lemmatize 적용한 후 TF-IDF 벡터화
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer(
    tokenizer= lemmatizer,      # 사용자 정의 토크나이저 적용
    stop_words= 'english',      # 영어 불용어 제거
    ngram_range= (1,2),         # 유니그램, 
    max_df= 0.85,               # 전체 문서 85% 등장시 흔한 단어이므로 제외
    min_df= 0.05                # 전체 문서 5% 등장시 
)

opinions_vecs = tfidf_vectorizer.fit_transform(document_df['opinions'])
print(opinions_vecs.toarray().shape)    # 문서수, 특성 수 
print(opinions_vecs)                    # 희소행렬 출력
````

### KMeans 군집화

````python
from sklearn.cluster import KMeans

kmeans = KMeans(
    n_clusters= 4,
    max_iter= 5000,
    random_state= 0
)

opinions_label = kmeans.fit_predict(opinions_vecs)
document_df['cluster'] = opinions_label
document_df
````

````python
# 클러스트 중신 키워드 (상위 TF-IDF ) 추출
centers = kmeans.cluster_centers_       # 각 클러스터별 중심 벡터
print(centers.shape)                    # (클러스터 수, 특성 수)

centroid_arg_idx = centers.argsort()[:, ::-1]   # 각 클러스터별 특성값 기준 내림차순하여, 인덱스 반환
top_20 = centroid_arg_idx[:, : 20]               # 클러스터별 상위 20개 특성 인덱스
feature_names = tfidf_vectorizer.get_feature_names_out()    # 이름
feature_names[top_20]       # 인덱스 -> 이름 매핑 출력
````

### 코사인 유사도 (cosine similarity)
- 코사인 유사도(cosine similarity)는 두 벡터가 이루는 각도로 “얼마나 비슷한 방향을 보느냐”를 재는 지표
- 크기(길이)보다 방향에 집중해서, 문서 벡터(TF-IDF) 비교에 자주 사용한다.


 # 
- 값 범위(보통 TF-IDF처럼 음수가 없으면): 0 ~ 1
- 1: 거의 같은 방향(매우 유사)
- 0: 직교(공통 특징 거의 없음)
- (일반 벡터에서는 -1도 가능: 정반대 방향)

````python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

doc_list = [
    'if you take the blue pill, the story ends',
    'if you take the blue pill, the stay in Wonderland',
    'if you take the blue pill, I show you how deep the rabbit hole goes',
]

tfidf_vectorizer = TfidfVectorizer()
doc_vects = tfidf_vectorizer.fit_transform(doc_list)    # 학습 + 문서 벡터화 (희소행렬)
print(doc_vects.shape)      # (문서 수, 단어 수)
print(sorted(tfidf_vectorizer.vocabulary_.items(), key = lambda x: x[1]))   # 단어사전은 인덱스 기준 정렬

doc_arr = doc_vects.toarray()       # 희소행렬 -> dense 배열 반환
cosine_similarity(doc_arr[2:], doc_arr) # 3번째 문서와 전체 문서 유사도
````
