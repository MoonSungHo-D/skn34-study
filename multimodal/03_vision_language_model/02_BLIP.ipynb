BLIP(Bootstrapping Language-Image Pre-training) 모델은 **이미지와 텍스트를 함께 이해하고 생성하는 멀티모달 AI 모델**이다.

기존 Vision-Language Pre-training(VLP) 모델들이 주로 이해(understanding) 혹은 생성(generation) 중 하나의 태스크에 특화되었던 것과 달리, BLIP은 **이해와 생성 모두에 유연하게 대응할 수 있도록 설계**되었다.


- **모델 구조**  
  BLIP은 **Multimodal Mixture of Encoder-Decoder(MED)** 구조를 사용한다.  
  - **Unimodal Encoder**: 이미지와 텍스트를 각각 독립적으로 인코딩하여 특징 벡터를 생성  
  - **Image-Text Encoder**: 이미지와 텍스트를 함께 인코딩해 두 modality의 관계를 학습  
  - **Image-Text Decoder**: 이미지를 참고해 텍스트(설명 등)를 생성

- **학습 목적(Pre-training Objectives)**  
  BLIP은 세 가지 주요 목적을 동시에 학습한다.
  1. **Image-Text Contrastive(ITC) Loss**: 이미지-텍스트 쌍의 임베딩이 서로 가까워지도록 대조적 학습
  2. **Image-Text Matching(ITM) Loss**: 이미지와 텍스트가 실제로 짝이 맞는지 분류
  3. **Language Modeling(LM) Loss**: 이미지를 참고해 자연스러운 텍스트를 생성하도록 학습

- **BLIP Task**

  ![](https://d.pr/i/X5x2cK+)

  1. Image-Text Retrieval: match여부 (이진분류)
  2. Image Captioning (텍스트생성)
  3. QuestionAnswering (시각적 질문/답변)


- **데이터 품질 개선(CapFilt)**  
  웹에서 수집한 대규모 이미지-텍스트 쌍에는 노이즈가 많기 때문에,  
  - **Captioner**: 이미지를 기반으로 합성 캡션 생성  
  - **Filter**: 노이즈가 많은 캡션을 자동으로 제거  
  이 과정을 통해 **고품질 데이터셋**을 구축하여 성능을 높인다.

- **적용 분야**  
  BLIP은 이미지-텍스트 검색, 이미지 캡셔닝, VQA(시각적 질의응답) 등 다양한 비전-언어 태스크에서 우수한 성능을 보인다.  
  또한, 비디오-언어 태스크 등으로의 **제로샷 전이** 성능도 뛰어나다.

- **기술적 특징**  
  - 이미지 인코더로 Vision Transformer(ViT) 사용  
  - 텍스트 인코더/디코더는 BERT 기반으로, Self-Attention 레이어를 제외한 파라미터를 공유하여 효율성을 높임

정리하면, **BLIP은 다양한 비전-언어 태스크에서 뛰어난 성능과 유연성을 가진 멀티모달 사전학습 모델**로, 고품질 데이터 구축과 효율적인 아키텍처 설계가 핵심이다.

```python
import requests
from PIL import Image
from transformers import BlipProcessor, BlipForConditionalGeneration

processor = BlipProcessor.from_pretrained('Salesforce/blip-image-captioning-base')
model = BlipForConditionalGeneration.from_pretrained('Salesforce/blip-image-captioning-base')
```

```python
img_url = 'https://storage.googleapis.com/sfr-vision-language-research/BLIP/demo.jpg'
raw_image = Image.open(requests.get(img_url, stream=True).raw).convert('RGB')
```

```python
text = 'a photograph of'
inputs = processor(raw_image, text, return_tensors = 'pt')  # 이미지 + 텍스트를 모델 입력 텐서로 변환
print(inputs['pixel_values'].shape)     # 이미지 텐서 (batch, channel, height, width)
print(inputs['input_ids'].shape)        # 텍스트 토큰 ID (batch, seq_len)
print(inputs['attention_mask'].shape)   # 패딩 마스크 (batch, seq_len)
```

```python
output = model.generate(**inputs)
print(output)
print(processor.decode(output[0], skip_special_tokens=True))
```

```python
inputs = processor(raw_image, return_tensors = 'pt')
print(inputs)

output = model.generate(**inputs)
print(processor.decode(output[0], skip_special_tokens=True))
```
