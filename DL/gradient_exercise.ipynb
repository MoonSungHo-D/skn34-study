```python
def numerical_diff(f, x, h=1e-4):
    return (f(x + h) - f(x - h)) / (2 * h)

def function(x):
    return x**2

print(numerical_diff(function, 3.0))
```

수치 미분은 조금 움직였을 때 값이 얼마나 변하는지 기울기를 추정할 때 사용한다.

### 02-06-02. 수치 미분의 한계

- 수치 미분은 **근사값**을 구하는 방법이므로, 정확한 해석적 미분값과 오차가 발생할 수 있다.
- 미분을 구하는 데 사용되는 **h 값의 선택이 중요**하다.
    - h 값이 너무 크면 근사 오차가 증가한다.
    - h 값이 너무 작으면 부동소수점 연산 오차가 커질 수 있다.

### 02-06-04. 편미분과 기울기 벡터

- 다변수 함수 $f(x, y)$에 대해, 특정 변수에 대한 편미분을 각각 계산할 수 있다. 이를 벡터 형태로 나타낸 것을 **기울기 벡터(Gradient Vector)** 라고 한다.

```python
def partial_diff(f, x, y, var='x', h=1e-4):
    if var == 'x':
        return (f(x + h, y) - f(x - h, y)) / (2 * h)
    else:
        return (f(x, y + h) - f(x, y - h)) / (2 * h)

def function(x, y):
    return x**2 + y**2

print(partial_diff(function, 3.0, 2.0, 'x'))
print(partial_diff(function, 3.0, 2.0, 'y'))
```

편미분은 다변화 함수에서 특정 변수 하나가 결과에 미치는 영향을 수식으로 나타낸것  

자동 미분은 
1. 모든 연산을 계산 그래프로 표현
2. 각 연산 노드에서 국소 미분 계산
3. 연쇄 법칙을 사용해 각 가중치에 대한 편미분 값을 계산

결론적으로 자동미분은 그 편미분 값을 계산 그래프와 연쇄 법칙을 이용해서 컴퓨터가 자동으로 계산해주는 기법

## 02-07. 기울기

### 02-07-01. 경사 하강법

- 경사 하강법(Gradient Descent)은 손실 함수의 기울기를 이용해 최적의 가중치를 찾아가는 방법이다.
- 특정 매개변수(예: 가중치 `w`)에 대해 손실 함수의 기울기를 구하고, 기울기의 반대 방향으로 이동하며 최적점을 찾는다.
- 학습률(learning rate, `lr`)이 너무 크면 최적점을 지나칠 수 있고, 너무 작으면 수렴 속도가 느려진다.

```python
def gradient_descent(f, init_x, lr=0.1, step_num=100):
    x = init_x
    for _ in range(step_num):
        grad = numerical_diff(f, x)
        x -= lr * grad
    return x

def function(x):
    return x**2

print(gradient_descent(function, 10.0))
```

손실을 최소화하기 위해서 손실의 반대반향으로 가중치를 수정한다.

$w \leftarrow w - \eta \frac{\partial L}{\partial w}$

w : 현재 가중치  
<- 새 값으로 업데이트 (:=)  
n (eta) 학습률  

∂L / ∂w :가중치 w가 손실함수 L에 미치는 영향(편미분, 기울기)  
- 는 손실을 줄이기 위해 기울기의 반대방향으로 이동  

결론적으로 현재의 가중치에서 손실을 가장 빠르게 증가시키는 방향의 반대 방향으로 학습률만큼 이동하여 가중치 갱신(손실을 줄이는 가장 효율적인 이동 방향)
