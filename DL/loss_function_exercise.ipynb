### 02-05-02. 평균 제곱 오차

- 평균 제곱 오차(Mean Squared Error, MSE)는 회귀 문제에서 주로 사용하는 손실 함수로, 실제값과 예측값의 차이를 제곱하여 평균을 낸 값이다.

```python
import numpy as np

def mean_squared_error(y_true, y_pred):
    return np.mean((y_true - y_pred) ** 2)

# 예제 데이터
y_true = np.array([3.0, -0.5, 2.0, 7.0])
y_pred = np.array([2.5, 0.0, 2.0, 8.0])

mse = mean_squared_error(y_true, y_pred)
print(f"MSE: {mse}")
```

### 02-05-03. 평균 절대 오차 (MAE, Mean Absolute Error)

- 평균 절대 오차는 실제값과 예측값의 차이의 절대값을 평균 낸 값이다.

```python
def mean_absolute_error(y_true, y_pred):
    return np.mean(np.abs(y_true - y_pred))

mae = mean_absolute_error(y_true, y_pred)
print(f"MAE: {mae}")
```

### 02-05-04. Huber 손실 (Huber Loss)

- Huber 손실은 MSE와 MAE의 장점을 결합한 손실 함수이다.
- 작은 오차에는 MSE처럼 작용하지만, 특정 임계값(delta) 이상에서는 MAE처럼 작용하여 이상치에 덜 민감하다.

```python
# Huber 손실 함수 : 작은 오차는 MSE, 큰 오차는 MAE 처럼 처리하여 이상치에 덜 민감하게 만든 손실함수
def huber_loss(y_true, y_pred, delta=1.0):
    error = y_true - y_pred
    is_small_error = np.abs(error) <= delta
    squared_loss = 0.5 * error**2
    linear_loss = delta * (np.abs(error) - 0.5 * delta)
    return np.mean(np.where(is_small_error, squared_loss, linear_loss))

huber = huber_loss(y_true, y_pred)
print(f"Huber Loss: {huber}")
```

### 02-05-05. 교차 엔트로피 오차

- 교차 엔트로피 오차(Cross Entropy Error, CEE)는 분류 문제에서 주로 사용하는 손실 함수이다. 실제값이 특정 클래스에 속할 확률과 예측 확률을 비교하여 손실을 계산한다.

```python
def cross_entropy_error(y_true, y_pred):
    delta = 1e-7  # log(0) 방지를 위한 작은 값
    return -np.sum(y_true * np.log(y_pred + delta))

# 예제 데이터 (One-hot encoding)
y_true = np.array([0, 0, 1])  # 실제 정답 (세 번째 클래스)
y_pred = np.array([0.2, 0.3, 0.5])  # 예측 확률

cee = cross_entropy_error(y_true, y_pred)
print(f"CEE: {cee}")
```

CEE 오차는   
정답클래스에 높은 점수를 주면 0에 가까워지고  
정답클래스에 낮은 점수를 주면 값이 커진다

- Softmax와 교차 엔트로피
    - 신경망의 출력층에서 Softmax 함수를 사용하여 확률 분포를 만든 후, CEE를 적용하여 모델 성능을 평가한다.

```python
def softmax(x):
    exp_x = np.exp(x - np.max(x))  # Overflow 방지
    return exp_x / np.sum(exp_x)

# 예제 데이터 (logits)
logits = np.array([2.0, 1.0, 0.1])
y_pred = softmax(logits)
print(f"Softmax Output: {y_pred}")

# 교차 엔트로피 오차 적용
y_true = np.array([0, 0, 1])
cee = cross_entropy_error(y_true, y_pred)
print(f"CEE: {cee}")
```

- BCE는 이진 분류 문제에서 사용하는 손실 함수로, 정답 클래스의 확률을 기반으로 오차를 계산한다.
    - Sigmoid 출력(0~1 확률)을 전제로 설계된 손실 함수이다.
    - 이진 분류에서 MSE보다 **수렴 속도와 성능이 일반적으로 우수**하다.
    - `nn.BCEWithLogitsLoss` 는 Sigmoid를 내부적으로 포함한 BCE로, 수치적으로 더 안정적이다.

### 02-05-07. 미니배치 학습 (Mini-batch Learning)

- 미니배치 학습(Mini-batch Learning)은 전체 데이터셋이 아닌 일부 데이터를 무작위로 선택하여 학습하는 방법이다.
- 전체 데이터를 한 번에 학습하는 것보다 계산량이 적고, SGD(확률적 경사 하강법)보다 안정적이다.

```python
def get_mini_batch(X, y, batch_size):
    idx = np.random.choice(
        len(X),         # 전체 인덱스
        batch_size,     # batch size 만큼
        replace=False)  # 중복 없이
    return X[idx], y[idx]

X = np.arange(100).reshape(50, 2)
y = np.arange(50)

X_batch, y_batch = get_mini_batch(X, y, batch_size=10)
print(f"미니배치 샘플: {X_batch}, {y_batch}")
```
