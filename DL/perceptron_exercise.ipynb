### 02-01-02. AND/NAND/OR 게이트

```python
import numpy as np

def perceptron(x1, x2, weights, bias):
    x = np.array([x1, x2])      # 입력값을 하나의 vector로 묶음
    w = np.array(weights)       # 가중치 리스트를 벡터로 변환
    
    # 퍼셉트론 계산 : 가중합(입력 벡터 *  가중치 벡터) + 편향
    return int(np.dot(x, w) + bias > 0)
    # return 1 if (w1*x1 + w2*x2 + b) > 0 else 0


# AND Gate
print("AND:", perceptron(1, 1, [0.5, 0.5], -0.7))

# NAND Gate (AND 연산을 반대로 만들기 위해 가중치 음수)
print("NAND:", perceptron(1, 1, [-0.5, -0.5], 0.7))


# OR Gate
print("OR:", perceptron(1, 0, [0.5, 0.5], -0.2))
```

AND, NAND, OR 연산은 단층 퍼셉트론에서 해결 가능

### 02-01-03. 가중치와 편향 구현

```python
def weighted_sum(inputs, weights, bias):
    return np.dot(inputs, weights) + bias

inputs = [1, 0]         # 입력
weights = [0.8, -0.5]   # 가중치
bias = 0.2              # 편향

output = weighted_sum(inputs, weights, bias)
print("Weighted Sum:", output)
```

### 02-01-05. 다층 퍼셉트론

```python
from sklearn.neural_network import MLPClassifier

# XOR 문제 해결
X = [[0, 0], [0, 1], [1, 0], [1, 1]]
y = [0, 1, 1, 0]

mlp = MLPClassifier(hidden_layer_sizes=(5,10),    # 은닉층 1개층, 뉴련 2개
                    activation='relu',          # relu 활성화 함수(가장 보편적으로 사용하는 활성화 함수)
                    max_iter=1000)              # epochs
mlp.fit(X, y)

print("Prediction:", mlp.predict(X))
```

activation : 은닉층에서 사용할 활성화 함수
- 입력에 비선형성을 추가해 복잡한 패턴 인식하는 함수

- ReLU 계열
    - relu : 가장 일반적으로 사용, 학습 빠름  (음수 허용을 안함)
    - Leaky ReLU : 음수 영역에 작은 기울기 허용 (죽은 ReLU 문제 완화시킨 함수)
    - ELU, GELU : Transformer 계열에서 사용

- tanh : 출력 범위가 -1 ~ 1 로 나오는게 의미 있을때 사용
    - 입력/ 출력이 0 을 중심으로 대칭
    - RNN 계열(LSTM, GRU)

- logistic : 시그모이드 함수(초기 신경망)


출력층에서 사용하는 활성화 함수와 구별을 잘해야한다.
- 이진분류 : sigmoid (출력범위 0 ~ 1) -> 확률로 해석
- 다중분류 : softmax (출력범위 0 ~ 1, 총합이 1) -> 각 클래스에 대한 확률 분포
- 회귀 : linear(없음) -> 예측값 그대로 사용
