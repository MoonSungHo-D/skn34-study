```python
import torch
import torch.nn.functional as F
```

```python
y_true = torch.tensor([3.0, 5.0, 7.0])
y_pred = torch.tensor([2.5, 5.5, 10.0])
```

```python
mse_loss = F.mse_loss(y_true, y_pred)           # MSE
mae_loss = F.l1_loss(y_true, y_pred)            # MAE
huber_loss = F.huber_loss(y_true, y_pred)       # Huber
    
print(f"{mse_loss.item():.4f}")
print(f"{mae_loss.item():.4f}")
print(f"{huber_loss.item():.4f}")
```

```python
y_pred_outlier = torch.tensor([2.5, 5.5, 20.0])

mse_loss = F.mse_loss(y_true, y_pred_outlier)           # MSE
mae_loss = F.l1_loss(y_true, y_pred_outlier)            # MAE
huber_loss = F.huber_loss(y_true, y_pred_outlier)       # Huber
    
print(f"{mse_loss.item():.4f}")
print(f"{mae_loss.item():.4f}")
print(f"{huber_loss.item():.4f}")
```

MSE Loss 는 제곱 연산을 하기 떄문에 큰 오차가 발생하면 손실값이 크게 증가한다.  
이상치가 많은 데이터의 경우 MAELoss 나 HuberLoss가 더 안정적으로 동작한다.
